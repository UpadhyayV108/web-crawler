```python
%%writefile app.py
import streamlit as st
import pandas as pd

import streamlit as st
import pandas as pd


data = [
    {"Name": "Dr. Alice Petri", "Title": "Director of Toxicology",
     "Company": "BioPharma.ai",
     "Location": "Boston", "HQ": "Boston", "Email": "alice@biopharma0.com",
     "Funding": "Series B", "Publications": 3},
    {"Name": "Dr.Oscar Lyons", "Title": "Head of Preclinical Safety",
     "Company": "MedTech Labs",
     "Location": "San Diego", "HQ": "San Diego", "Email": "john@medtech1.com",
     "Funding": "Series A", "Publications": 1},
]

df = pd.DataFrame(data)

def score_lead(row):
    score = 0
    if "Toxicology" in row["Title"] or "Safety" in row["Title"]:
        score += 30
    if row["Funding"] == "Series A":
        score += 20
    elif row["Funding"] == "Series B":
        score += 25
    if row["Location"] in ["Boston", "Cambridge", "San Diego"]:
        score += 10
    if row["Publications"] > 0:
        score += 40
    return score

df["Score"] = df.apply(score_lead, axis=1)

st.title("meds lead Dashboard")

city_filter = st.text_input("Filter by City")
if city_filter:
    df = df[df["Location"].str.contains(city_filter, case=False)]


st.dataframe(df.sort_values(by="Score", ascending=False))

st.download_button("Download CSV", df.to_csv(index=False), "leads.csv", "text/csv")
```
